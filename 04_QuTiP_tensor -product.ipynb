```python
from qutip import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
omega = 2*np.pi

H = 0.5 * omega * sigmax()
```

```python
psi0 = basis(2,0)
```

```python
tlist = np.linspace(0,5,100)       #sayad ye time hain jo tum...
```

```python
e_ops = [sigmaz()]     #is code ko aur samjahna ⟨σz​⟩ so ehre we are fidnign the excepation value sof sigma z
```

```python
result = sesolve(H, psi0, tlist, e_ops=e_ops)
```
